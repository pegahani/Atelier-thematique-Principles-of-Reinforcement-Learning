```python
import numpy as np, matplotlib, gymnasium as gym, torch, stable_baselines3 as sb3
for m in [np, matplotlib, gym, torch, sb3]:
    print(f"{m.__name__:20s} {m.__version__}")
print("✅ packages OK")
```

```python
# Classic and Box2D environments
for env_id in ["FrozenLake-v1", "CartPole-v1", "LunarLander-v3"]:
    env = gym.make(env_id)
    obs, info = env.reset(seed=0)
    obs, reward, terminated, truncated, info = env.step(env.action_space.sample())
    env.close()
    print(f"{env_id:16s} observation space: {env.observation_space}")
print("✅ environments OK (including Box2D)")
```

```python
# PyTorch on CPU
x = torch.randn(256, 8)
net = torch.nn.Sequential(torch.nn.Linear(8, 64), torch.nn.ReLU(), torch.nn.Linear(64, 4))
loss = net(x).pow(2).mean()
loss.backward()
print("device:", "cuda" if torch.cuda.is_available() else "cpu", "| threads:", torch.get_num_threads())
print("✅ PyTorch OK")
```

```python
# A tiny Stable-Baselines3 training run (about 10-20 seconds)
import time
from stable_baselines3 import PPO
start = time.time()
model = PPO("MlpPolicy", "CartPole-v1", n_steps=512, verbose=0, seed=0, device="cpu")
model.learn(total_timesteps=10_000)
fps = 10_000 / (time.time() - start)
print(f"training speed: {fps:.0f} steps/s")
print("✅ Stable-Baselines3 OK")
```
